```python
import numpy as np
```

```python
# 1. Create an array containing 10, 20, 30, 40, 50
a = np.array([10, 20, 30, 40, 50])
print(a)
```

```python
# 2. Create an array containing integers from 1 to 20
a = np.arange(1, 21)
print(a)
```

```python
# 3. Create an array of 10 zeros
a = np.zeros(10)
print(a)
```

```python
# 4. Create a 3 x 4 array filled with ones
a = np.ones((3, 4))
print(a)
```

```python
# 5. Create a 2 x 3 array where every element is 7
a = np.full((2, 3), 7)
print(a)
```

```python
# 6. Create 10 equally spaced values between 0 and 100
a = np.linspace(0, 100, 10)
print(a)
```

```python
# 7. Create a 4 x 4 identity matrix
a = np.eye(4)
print(a)
```

```python
# 8. Create even numbers from 2 to 50
a = np.arange(2, 51, 2)
print(a)
```

```python
# 9(a). Reshape the array into a 3 x 4 matrix
a = np.arange(1, 13)
b = a.reshape(3, 4)
print(b)
```

```python
# 9(b). Reshape the array into a 4 x 3 matrix
a = np.arange(1, 13)
b = a.reshape(4, 3)
print(b)
```

```python
# 9(c). Reshape the array into a 2 x 2 x 3 array
a = np.arange(1, 13)
b = a.reshape(2, 2, 3)
print(b)
```
