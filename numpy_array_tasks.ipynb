```python
import numpy as np
```

```python
a = np.array([10, 20, 30, 40, 50])
print(a)
```

```python
a = np.arange(1, 21)
print(a)
```

```python
a = np.zeros(10)
print(a)
```

```python
a = np.ones((3, 4))
print(a)
```

```python
a = np.full((2, 3), 7)
print(a)
```

```python
a = np.linspace(0, 100, 10)
print(a)
```

```python
a = np.eye(4)
print(a)
```

```python
a = np.arange(2, 51, 2)
print(a)
```

```python
a = np.arange(1, 13)
b = a.reshape(3, 4)
print(b)
```

```python
a = np.arange(1, 13)
b = a.reshape(4, 3)
print(b)
```

```python
a = np.arange(1, 13)
b = a.reshape(2, 2, 3)
print(b)
```
