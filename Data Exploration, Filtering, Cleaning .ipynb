```python
#Calculate delivery duration
orders['order_delivered_customer_date'] = pd.to_datetime(orders['order_delivered_customer_date'])
orders['delivery_duration'] = (orders['order_delivered_customer_date'] - orders['order_purchase_timestamp']).dt.days
print(orders[['order_id', 'delivery_duration']])
```

```python
#Calculate the difference between estimated and actual delivery dates
orders['order_estimated_delivery_date'] = pd.to_datetime(orders['order_estimated_delivery_date'])
orders['delivery_difference'] = (orders['order_delivered_customer_date'] - orders['order_estimated_delivery_date']).dt.days
print(orders[['order_id', 'delivery_difference']])
```

```python
#Create a delivery status column – Early, On Time, Delayed
def check_status(days):
    if pd.isna(days):
        return 'Not Delivered'
    elif days < 0:
        return 'Early'
    elif days == 0:
        return 'On Time'
    else:
        return 'Delayed'

orders['delivery_status'] = orders['delivery_difference'].apply(check_status)

print(orders[['order_id', 'delivery_status']])
```

```python
#Rename columns using meaningful names
orders.rename(columns={'order_purchase_timestamp': 'purchase_date',
    'order_delivered_customer_date': 'actual_delivery_date',
    'order_estimated_delivery_date': 'estimated_delivery_date'}, inplace = True)

print(orders.columns)
```
