```python
#Pandas Practice Questions
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(r"/content/insurance.csv")
```

```python
#Display the first 10 and last 10 records.
print(df.head(10))
print(df.tail(10))
```

```python
#Find the number of rows and columns.
print("Rows: ",df.shape[0])
print("Cols: ",df.shape[1])
```

```python
#Display column names and data types.
print(df.columns)
print(df.dtypes)
```

```python
#Find missing values in each column.
print(df.isnull().sum())
```

```python
#Check duplicate records.
print(df.duplicated().sum())
print(df[df.duplicated()])
```

```python
#Generate statistical summaries.
print(df.describe())
```

```python
#Find unique values in the region column.
print(df['region'].unique())
```

```python
#Count customers by gender.
print(df['sex'].value_counts())
```

```python
#Count smokers and non-smokers.
print(df['smoker'].value_counts())
```

```python
#Sort customers by insurance charges in descending order.
sorted = df.sort_values(by=['charges'], ascending=False)
print(sorted.head)
```

```python
#Extract customers aged above 50.
aged_customers = df[df['age']>50]
print(aged_customers)
print(len(aged_customers))
```

```python
#Filter customers with BMI above 30.
over_bmi = df[df['bmi']>30]
print(over_bmi)
print(len(over_bmi))
```

```python
#Extract female smokers.
female_smokers = df[(df['sex'] == 'female') & (df['smoker'] == 'yes')]
print(female_smokers)
print(len(female_smokers))
```

```python
#Find customers from the southeast region.
southeast = df[df['region'] == 'southeast']
print(southeast)
print(len(southeast))
```

```python
#Filter customers whose charges are between $10,000 and $30,000.
filtered = df[(df['charges'] >= 10000) & (df['charges'] <= 30000)]
print(filtered)
print(len(filtered))
```

```python
#Create a new column called age_group.
df['age_group'] = pd.cut(df['age'], bins=[0,29,50,100], labels = ['Young', 'Adult', 'Senior'])
print(df[['age', 'age_group']])
```

```python
#Create a new column called bmi_category.
df['bmi_category'] = pd.cut(df['bmi'], bins=[0, 18.5, 25, 30, float('inf')], labels = ['Underweight', 'Normal', 'Overweight', 'Obese'], right = False)
print(df[['bmi', 'bmi_category']])
```

```python
#Convert smoker values into numerical values (yes = 1, no = 0).
df['smoker'] = df['smoker'].map({'yes': 1,'no': 0})
print(df[['smoker']])
```

```python
#Create a new column called high_cost_customer.
df['high_cost_customer'] =df['charges']>20000
print(df[['charges','high_cost_customer']])
```

```python
#Rename the columns into uppercase.
df = df.rename(columns=str.upper)
print(df.columns)
```

```python
#Calculate average charges by gender.
result = df.groupby('SEX')['CHARGES'].mean()
print(result)
```

```python
#Find total charges by region.
result = df.groupby('REGION')['CHARGES'].sum()
print(result)
```
