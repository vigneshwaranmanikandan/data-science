```python
import pandas as pd
```

```python
h = pd.read_csv(r"/content/Housing.csv")
```

```python
h.columns
```

```python
x=h.iloc[:, 1:]
y=h['price']
```

```python
x
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
l = LabelEncoder()
cols_to_encode = ['mainroad', 'guestroom', 'basement', 'hotwaterheating', 'airconditioning', 'prefarea', 'furnishingstatus']
for col in cols_to_encode:
    h[col] = l.fit_transform(h[col])
```

```python
h.info()
```

```python
h.corr()
```

```python
h.head()
```

```python
x = h.iloc[:, 1:]
y = h['price']
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)
y_pred
```

```python
from sklearn.metrics import mean_squared_error, r2_score
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)
r2
```

```python
from sklearn.linear_model import Lasso, Ridge
```

```python
l1 = Lasso()
l1.fit(x_train, y_train)
```

```python
l1_pred = l1.predict(x_test)
l1_pred
```

```python
l1.coef_
```

```python
eval = r2_score(y_test, l1_pred)
eval
```
