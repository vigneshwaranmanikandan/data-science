```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(r"/content/insurance.csv")
```

```python
print(df.head())
```

```python
#Calculate mean, median, mode and standard deviation of age, bmi and charges.
print("Age Mean: ",round(df['age'].mean(),2))
print("Age Median: ",df['age'].median())
print("Age Mode: ", df['age'].mode())
print("Age S.D.: ", round(df['age'].std(),2))
print("")
print("bmi Mean: ",round(df['bmi'].mean(),2))
print("bmi Median: ",df['bmi'].median())
print("bmi Mode: ", df['bmi'].mode())
print("bmi S.D.: ", round(df['bmi'].std(),2))
print("")
print("charges Mean: ",round(df['charges'].mean(),2))
print("charges Median: ",df['charges'].median())
print("charges Mode: ", df['charges'].mode())
print("charges S.D.: ", round(df['charges'].std(),2))
```

```python
#Plot histograms for all three numerical columns.
import matplotlib.pyplot as plt

plt.figure(figsize=(18, 5))
plt.subplot(1, 3, 1)
plt.hist(df['age'], bins=10)
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')

plt.subplot(1, 3, 2)
plt.hist(df['bmi'], bins=10)
plt.title('BMI Distribution')
plt.xlabel('BMI')
plt.ylabel('Frequency')

plt.subplot(1, 3, 3)
plt.hist(df['charges'], bins=10)
plt.title('Charges Distribution')
plt.xlabel('Charges')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()
```

```python
#Calculate skewness and kurtosis.
columns = ['age', 'bmi', 'children', 'charges']
for col in columns:
  print(col)
  print("Skewness: ", df[col].skew())
  print("Kurtosis: ", df[col].kurt())
  print()
```

```python
#Identify which columns appear normally distributed.
import scipy.stats as stats

plt.figure(figsize=(12, 8))

for i, col in enumerate(columns, 1):
  plt.subplot(2,2,i)
  stats.probplot(df[col], dist='norm', plot=plt)
  plt.title(f'Q-Q Plot of {col}')

plt.tight_layout()
plt.show()
```

```python
#Identify whether charges are positively or negatively skewed.
print("Charges Skewness: ", df['charges'].skew())
plt.hist(df['charges'],bins=30)
plt.title('Charges Distribution')
plt.xlabel('Index')
plt.ylabel('Charges')
plt.show()
```

```python
#Create boxplots and identify outliers.
plt.figure(figsize=(12,8))

for i, col in enumerate(columns,1):
  plt.subplot(2,2,i)
  plt.boxplot(df[col])
  plt.title(f'Boxplot of {col}')

plt.tight_layout()
plt.show()
```

```python
#Compare mean and median. Why are they different?
print("Mean and Median are different because mean sums all the values and find average, while median finds the exact center value among the values in ascending order")
print("Age Mean: ",round(df['age'].mean(),2))
print("Age Median: ",df['age'].median())
print("In age, the mean and median values are almost similar")
print()
print("bmi Mean: ",round(df['bmi'].mean(),2))
print("bmi Median: ",df['bmi'].median())
print("In bmi, the mean and median values are almost similar")
print("charges Mean: ",round(df['charges'].mean(),2))
print("charges Median: ",df['charges'].median())
print("In charges, the mean and median values differes more. Because, there maybe more outliers (extreme) values in charges")
```

```python
#Plot a normal distribution curve.
from scipy.stats import norm,binom
x = np.linespace(-4,4,1000)
y = norm.pdf(x, loc=0, scale=1)
plt.figure(figsize=(9,5))
plt.plot(x,y)
plt.title("Normal Distribution")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.grid(axis = 'y')
plt.show()
```

```python
#Compare the BMI distribution with a theoretical normal distribution.
bmi = df['bmi']

mean_bmi = np.mean(bmi)
std_bmi = np.std(bmi)
x = np.linspace(bmi.min(), bmi.max(), 1000)
y = norm.pdf(x, mean_bmi, std_bmi)
plt.figure(figsize=(10,5))
plt.hist(bmi, bins=30, density=True, alpha=0.6, label='Actual BMI distribution')
plt.plot(x,y,linewidth=2,label='Theoretical Normal Distribution')
plt.title("BMI vs Theoretical Normal Distribution")
plt.xlabel("BMI")
plt.ylabel("Density")
plt.legend()
plt.grid(axis = 'y')
plt.show()
```

```python
#Calculate the probability that a randomly selected customer has BMI above 30.
total_customers = len(df)
high_bmi = np.sum(df['bmi'] > 30)
probability = high_bmi / total_customers
print("Customers with bmi above 30:", high_bmi)
print("Total customers:", total_customers)
print("Probability:", probability)
print("Percentage:", probability * 100)
```

```python
#Generate a Poisson distribution with a chosen event rate and visualize it.
from scipy.stats import poisson

lam = 4
data = np.random.poisson(lam=lam,size=1000)
plt.figure(figsize=(9, 5))
plt.hist(data,bins=np.arange(-0.5, max(data) + 1.5, 1),density=True,edgecolor='black')
plt.title("Poisson Distribution - 1000 Samples")
plt.xlabel("Number of Events")
plt.ylabel("Probability")

plt.grid(axis='y')
plt.show()
```

```python
#Theoretical normal distribution
x = np.arange(0,15)
y=poisson.pmf(x,lam)
plt.figure(figsize=(9,5))
plt.bar(x,y,edgecolor='black')
plt.title("Theoretical Poisson Distribution")
plt.xlabel("Number of Events")
plt.ylabel("Probability")
plt.grid(axis='y')
plt.show()
```

```python
#Compare uniform and normal distributions visually.
normal_data = np.random.normal(loc=50,scale=10,size=1000)
uniform_data = np.random.uniform(low=20,high=80,size=1000)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].hist(normal_data,bins=30,edgecolor='black')
axes[0].set_title("Normal Distribution")
axes[0].set_xlabel("Values")
axes[0].set_ylabel("Frequency")

axes[1].hist(uniform_data,bins=30,edgecolor='black')
axes[1].set_title("Uniform Distribution")
axes[1].set_xlabel("Values")
axes[1].set_ylabel("Frequency")

plt.tight_layout()
plt.show()
```
