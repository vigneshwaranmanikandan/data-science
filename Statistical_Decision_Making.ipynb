```python
import pandas as pd
df = pd.read_csv(r"/content/insurance.csv")
```

```python
#Data preparation
required_columns =['sex', 'bmi', 'smoker', 'region', 'charges']
data = df[required_columns].copy()
data['sex'] = data['sex'].astype(str).str.lower().str.strip()
data['smoker'] = data['smoker'].astype(str).str.lower().str.strip()
data['region'] = data['region'].astype(str).str.lower().str.strip()
print("Dataset shape:",data.shape)
print("\nMissing values:")
print(data.isnull().sum())

alpha = 0.05
```

```python
smokers = data[data['smoker'] == 'yes']['charges']
non_smokers = data[data['smoker'] == 'no']['charges']

from scipy.stats import ttest_ind
t_stat_1, p_value_1 = ttest_ind(smokers, non_smokers, equal_var=False)
difference_1 = smokers.mean()-non_smokers.mean()
print("\n Question 1: Smokers vs Non-Smokers Charges")
print("Smokers average: ", round(smokers.mean(),2))
print("Non-Smokers average: ", round(non_smokers.mean(),2))
print("Difference: ", round(difference_1,2))
print("T-stats: ", round(t_stat_1,2))
print("P Value: ", round(p_value_1, 2))

if p_value_1 < alpha:
  print("Result: Significant difference in charges")
else:
  print("Result: No statistically significant difference")
```

```python
males = data[data['sex'] == 'male']['bmi']
females = data[data['sex'] == 'female']['bmi']

t_stat_2, p_value_2 = ttest_ind(males, females, equal_var=False)
difference_2 = males.mean()-females.mean()
print("\n Question 2: Males vs Females BMI")
print("Males average: ", round(males.mean(),2))
print("Females average: ", round(females.mean(),2))
print("Difference: ", round(difference_2,2))
print("T-stats: ", round(t_stat_2,2))
print("P Value: ", round(p_value_2, 2))

if p_value_1 < alpha:
  print("Result: Significant difference in average bmi")
else:
  print("Result: No statistically significant difference")
```

```python
from scipy.stats import stats,t
import numpy as np
sample_100 = data['charges'].sample(n=100,random_state=42)

sample_mean = sample_100.mean()
sample_std = sample_100.std()
sample_n = len(sample_100)

confidence = 0.95
alpha_ci = 1 - confidence

critical_value = t.ppf(1 - alpha_ci / 2,df=sample_n - 1)
standard_error = sample_std / np.sqrt(sample_n)
margin_error = critical_value * standard_error

lower_100 = sample_mean - margin_error
upper_100 = sample_mean + margin_error

population_mean = data['charges'].mean()

print("\nQuestion 5: Sample of 100 customers")
print("Sample mean:", round(sample_mean, 2))
print("95% confidence interval:")
print(round(lower_100, 2), "to", round(upper_100, 2))

print("Actual dataset mean:", round(population_mean, 2))
print("Margin of error:", round(margin_error, 2))
```
