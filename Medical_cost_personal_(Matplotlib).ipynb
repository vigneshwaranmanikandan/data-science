```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv(r"/content/insurance.csv")
```

```python
#Create a line plot showing age versus insurance charges.
plt.figure(figsize=(10, 5))
plt.plot(df['age'], df['charges'])
plt.title("Age vs Insurance Charges")
plt.xlabel("Age")
plt.ylabel("Insurance Charges ($)")
plt.grid(True)
plt.show()
```

```python
#Create a bar chart showing customer count by region.
region_count = df['region'].value_counts()
plt.figure(figsize=(8, 5))
plt.bar(region_count.index, region_count.values)
plt.title("Customer Count by Region")
plt.xlabel("Region")
plt.ylabel("Customer Count")
plt.grid(axis='y')
plt.show()
```

```python
#Create a pie chart showing smoker versus non-smoker distribution.
smoker_count = df['smoker'].value_counts()
plt.figure(figsize=(7, 7))
plt.pie(smoker_count.values,labels=smoker_count.index,autopct='%1.1f%%',startangle=90)

plt.title("Smoker Distribution")
plt.show()
```

```python
#Create a histogram of customer ages.
plt.figure(figsize=(8, 5))
plt.hist(df['age'], bins=10)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.grid(axis='y')
plt.show()
```

```python
#Create a histogram of insurance charges.
plt.figure(figsize=(8, 5))
plt.hist(df['charges'], bins=20)
plt.title("Insurance Charges Distribution")
plt.xlabel("Insurance Charges")
plt.ylabel("Frequency")
plt.grid(axis='y')
plt.show()
```

```python
#Create a scatter plot of BMI versus charges.
plt.figure(figsize=(8, 5))
plt.scatter(df['bmi'], df['charges'])
plt.title("BMI vs Insurance Charges")
plt.xlabel("BMI")
plt.ylabel("Insurance Charges")
plt.grid(True)
plt.show()
```

```python
#Create a horizontal bar chart showing average charges by gender.
gender_charges = df.groupby('sex')['charges'].mean()
plt.figure(figsize=(8, 5))
plt.barh(gender_charges.index, gender_charges.values)
plt.title("Average Charges by Gender")
plt.xlabel("Average Charges")
plt.ylabel("Gender")

plt.grid(axis='x')
plt.show()
```

```python
#Create a box plot showing charges distribution.
plt.figure(figsize=(8, 5))
plt.boxplot(df['charges'])
plt.title("Insurance Charges Distribution")
plt.ylabel("Insurance Charges")
plt.grid(axis='y')
plt.show()
```

```python
#Compare smoker and non-smoker charges using a bar chart.
smoker_charges = df.groupby('smoker')['charges'].mean()
plt.figure(figsize=(8, 5))
plt.bar(smoker_charges.index, smoker_charges.values)
plt.title("Average Charges: Smokers vs Non-Smokers")
plt.xlabel("Smoker Status")
plt.ylabel("Average Charges")
plt.grid(axis='y')
plt.show()
```

```python
#Create separate histograms for male and female customers.
male = df[df['sex'] == 'male']
female = df[df['sex'] == 'female']
plt.figure(figsize=(10, 5))
plt.hist(male['age'], bins=10,alpha=0.5, label='Male')
plt.hist(female['age'], bins=10,alpha=0.5, label='Female')
plt.title("Age Distribution by Gender")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.legend()
plt.grid(axis='y')
plt.show()
```

```python
#Visualize average charges by region.
region_charges = df.groupby('region')['charges'].mean()
plt.figure(figsize=(9, 5))
plt.bar(region_charges.index, region_charges.values)
plt.title("Average Insurance Charges by Region")
plt.xlabel("Region")
plt.ylabel("Average Charges ($)")
plt.grid(axis='y')
plt.show()
```

```python
#Compare BMI distribution across different regions using box plots.
regions = df['region'].unique()
bmi_data = [df[df['region'] == region]['bmi'] for region in regions]
plt.figure(figsize=(9, 5))
plt.boxplot(bmi_data, labels=regions)
plt.title("BMI Distribution Across Regions")
plt.xlabel("Region")
plt.ylabel("BMI")

plt.grid(axis='y')
plt.show()
```

```python
#Create a scatter plot of age versus charges, using different colors for smokers and non-smokers.
smokers = df[df['smoker'] == 'yes']
non_smokers = df[df['smoker'] == 'no']
plt.figure(figsize=(10, 6))
plt.scatter(non_smokers['age'],non_smokers['charges'],label='Non-Smoker',alpha=0.6)
plt.scatter(smokers['age'],smokers['charges'],label='Smoker',alpha=0.6)
plt.title("Age vs Insurance Charges by Smoking Status")
plt.xlabel("Age")
plt.ylabel("Insurance Charges")
plt.legend()
plt.grid(True)
plt.show()
```

```python
#Create a subplot containing age distribution, BMI distribution and charges distribution.
fig,axes=plt.subplots(1,3,figsize=(15,5))
axes[0].hist(df['age'], bins=10)
axes[0].set_title("Age Distribution")
axes[0].set_xlabel("Age")
axes[0].set_ylabel("Frequency")

axes[1].hist(df['bmi'], bins=10)
axes[1].set_title("BMI Distribution")
axes[1].set_xlabel("BMI")
axes[1].set_ylabel("Frequency")

axes[2].hist(df['charges'], bins=20)
axes[2].set_title("Charges Distribution")
axes[2].set_xlabel("Charges ($)")
axes[2].set_ylabel("Frequency")

plt.tight_layout()
plt.show()
```

```python
#Add data labels on top of bars.
region_count = df['region'].value_counts()
plt.figure(figsize=(9, 5))
bars = plt.bar(region_count.index, region_count.values)
plt.bar_label(bars)
plt.title("Customer Count by Region")
plt.xlabel("Region")
plt.ylabel("Customer Count")

plt.show()
```

```python
#Sort the bar chart by average charges.
region_charges = (df.groupby('region')['charges'].mean().sort_values(ascending=False))
plt.figure(figsize=(9, 5))
bars = plt.bar(region_charges.index,region_charges.values)
plt.bar_label(bars, fmt='%.0f')
plt.title("Average Charges by Region")
plt.xlabel("Region")
plt.ylabel("Average Charges")

plt.grid(axis='y')
plt.show()
```

```python
#Add annotations to highlight the highest insurance charge.
max_index = df['charges'].idxmax()
max_age = df.loc[max_index, 'age']
max_charge = df.loc[max_index, 'charges']
plt.figure(figsize=(10, 6))
plt.scatter(df['age'], df['charges'])
plt.annotate(f"Highest: {max_charge:.2f}",xy=(max_age, max_charge),xytext=(max_age - 15, max_charge - 10000),arrowprops=dict(arrowstyle='->'))
plt.title("Age vs Insurance Charges")
plt.xlabel("Age")
plt.ylabel("Insurance Charges")
plt.grid(True)
plt.show()
```

```python
#Create a chart comparing average charges across age groups.
df['age_group'] = pd.cut(df['age'],bins=[17, 29, 39, 49, 59, 64],labels=['18-29', '30-39', '40-49', '50-59', '60-64'])
age_charges = df.groupby('age_group')['charges'].mean()
plt.figure(figsize=(9, 5))
bars = plt.bar(age_charges.index.astype(str),age_charges.values)
plt.bar_label(bars, fmt='%.0f')
plt.title("Average Insurance Charges by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Average Charges")

plt.grid(axis='y')
plt.show()
```
